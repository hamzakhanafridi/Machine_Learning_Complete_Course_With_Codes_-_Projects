# creating and comparing bars on 2 datasets

```python
oscar_movie=[
    'dark night',
    'hurt locker',
    'kings speech',
    'argo'
]
oscar_revinue=[1005,170,450,280]

non_oscar_movie=[
    'dark',
    'avatar',
    'kings',
    'argo king'
]
non_oscar_revinue=[550,3000,300,480]
year=[1998, 2004,2011,2019]

import numpy as np 
x=np.arange(len(year))
width=0.35
plt.bar(x-width/2,oscar_revinue ,width, label='Oscar movie');
plt.bar(x+width/2,non_oscar_revinue ,width, label='Non_Oscar movie');
plt.title('Oscar Vs Non Oscar Movie revinue')
plt.xlabel('Year')
plt.ylabel('Revinue in M$')
plt.legend()

plt.xticks(x, year)
```

```python
# horizontal bar char

oscar_movie=[
    'dark night',
    'hurt locker',
    'kings speech',
    'argo'
]
oscar_revinue=[1005,170,450,280]
plt.barh(oscar_movie,oscar_revinue)
plt.title('movies vs revinue')
plt.xlabel('revinue')
plt.ylabel('movies')
```

```python
# Scatter plot
person = [
    "Ali", "Ahmed", "Usman", "Hamza", "Bilal",
    "Saad", "Zain", "Hassan", "Omar", "Ayan",
    "Danish", "Talha", "Arham", "Rayan", "Fahad"
]

age = [18, 21, 25, 29, 32, 35, 38, 42, 45, 48, 52, 56, 60, 65, 70]

bp = [112, 118, 121, 125, 128, 130, 133, 138, 140, 145, 148, 152, 155, 160, 165]

plt.scatter(age, bp, color='red')
plt.title('Age vs BP')
plt.xlabel('Age')
plt.ylabel('BP')
plt.grid()
```

```python
colors=['green' if x< 135 else 'red' for x in bp]
# plt.scatter(age, bp, color=colors)
# plt.scatter(age, bp, cmap='plasma', c=bp)
plt.scatter(age, bp, s=bp, cmap='OrRd', c=bp , alpha=0.5)
plt.title('Age vs BP')
plt.xlabel('Age')
plt.ylabel('BP')
plt.colorbar(label='bp')
plt.grid()
```

```python
# Annotation (person ke text ko en points pe print karnaa)
colors=['green' if x< 135 else 'red' for x in bp]
plt.scatter(age, bp,cmap='OrRd', c=bp)

for i in range(len(person)):
    plt.annotate(person[i], xy=(age[i],bp[i]),xytext=(age[i], bp[i]+1))
plt.title('Age vs BP')
plt.xlabel('Age')
plt.ylabel('BP')

plt.xlim(min(age), max(age)+10)
plt.ylim(min(bp), max(bp)+5)
plt.colorbar(label='bp')
plt.grid()

plt.savefig('scatter_final.png')
```

# Scatter plot on Multiple datasets

```python
cities = [
    "Karachi", "Lahore", "Islamabad", "Peshawar", "Quetta",
    "Dubai", "London", "New York", "Tokyo", "Singapore",
    "Sydney", "Toronto", "Moscow", "Cairo", "Delhi",
    "Bangkok", "Paris", "Berlin", "Madrid", "Istanbul"
]

summer_temp = [
    34, 38, 32, 35, 31,
    41, 22, 29, 30, 32,
    25, 24, 18, 36, 35,
    33, 25, 24, 33, 29
]

winter_temp = [
    20, 14, 12, 10, 4,
    20, 8, 2, 7, 26,
    18, -2, -7, 19, 14,
    25, 7, 3, 9, 8
]

summer_humidity = [
    70, 55, 45, 50, 30,
    55, 75, 65, 75, 80,
    70, 60, 75, 45, 60,
    75, 65, 70, 40, 65
]

winter_humidity = [
    65, 60, 55, 65, 55,
    65, 82, 70, 55, 75,
    65, 75, 80, 55, 60,
    70, 85, 80, 75, 75
]
plt.scatter(summer_temp, summer_humidity, label='Summer')
plt.scatter(winter_temp, winter_humidity, label='winter')

plt.title('Summer and winter temp vs humidity')
plt.xlabel('Temperature')
plt.ylabel('humidity')
plt.legend()
```
