```python
# index and label base access of values
s=pd.Series([23,28,31], index=['Hamza','bob', 'tom'])
print(s)
```

```python
print(s['Hamza'])   # label base access of values  now a days for label wise access loc is used
print(s.loc['bob'])   # same label base access but using loc keyword
print(s.iloc[1])   # index base access of values  for index wise access iloc is used
```

```python
#   Dataframe and its creating methods   (3 methods)
# 1. creating it from dictionary of lsits
dc={
    'name': ['hamza','tom', 'bob'],
    'age': [27,31,55],
    'gpa':[3.9,3.3,2.9]
}

df=pd.DataFrame(dc)
df

#  2. creating it from lsit of lists
listt=[[1,2,3], [4,5,6]]
df_list=pd.DataFrame(listt, columns=['a','b','c'])
df_list

#  creating it from numpy array
import numpy as np
n=np.array([[1,3,4],[6,7,8]])
num_df=pd.DataFrame(n, columns=['A','B','C'])
num_df
```
