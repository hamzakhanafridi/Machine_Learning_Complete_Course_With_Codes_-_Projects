```python
class KNN_classifier:
    def __init__(self, k=3):
        self.k=k

    def _eucliden_dist(self, x1,x2):
        return np.sqrt(np.sum((x1-x2)**2))

    def fit(self,X,y):
        self.x_train= X
        self.y_train=y

    def _predict_one(self,x):
        distances = [self._eucliden_dist(x, x_train) for x_train in self.x_train]
    
        knn_indices = np.argsort(distances)[:self.k]
    
        knn_classes = [self.y_train[i] for i in knn_indices]
    
        majority_class = np.argmax(np.bincount(knn_classes))
    
        return majority_class

    def predict(self,X):
        y_pred=[self._predict_one(x) for x in X]
        return np.array(y_pred)
```

```python
x_train = np.array([
    [1, 2],
    [2, 3],
    [3, 3],
    [6, 5],
    [7, 7]
])

y_train = np.array([0, 0, 0, 1, 1])

x_test = np.array([
    [2, 2],
    [6, 6]
])

model = KNN_classifier()
model.fit(x_train, y_train)

y_pred = model.predict(x_test)

print(y_pred)     # 2 test point jes me 1 zero ayega and 1 one  jo ke ham ne code se pehle seaborn se be deka ta
```
