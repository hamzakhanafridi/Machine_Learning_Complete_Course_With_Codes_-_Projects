```python
import numpy as np
```

```python
class LinearReg:
    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.bias=None
        self.weight=None
        self.lr=learning_rate
        self.n_iter=n_iter

    def fit(self, x,y):

        #step 1
        m,n=x.shape
        self.bias=0
        self.weight=np.zeros(n)

        for i in range(self.n_iter):
            # Step 2
            y_pred=self.bias + np.dot(x, self.weight)   #self.bias is teta(0) and dot is dot product and self.weight is all tetas here which multiply with x
    
            # Step 3
            db=(1/m) * np.sum(y_pred - y)    # db mean derivative w.r.t teta not
            dw=(1/m) * np.dot(x.T, (y_pred - y))
    
            # Step 4 applying convergence theorem 
            self.bias -= self.lr * db
            self.weight -= self.lr * dw

    def predict(self, x):
        y_pred=self.bias + np.dot(x, self.weight)
        return y_pred
```

```python
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([2,4,6,8,10])

model = LinearReg()
model.fit(x,y)
y_pred=model.predict(x)
print(y_pred)
```

# Now doing Linear regression with OLS(Ordinary Least Square)

```python
class LinearRegOLS:
    def __init__(self):
        self.bias=None
        self.weight=None

    def fit(self, x,y):
        m,n=x.shape
        x_b=np.c_[np.ones((m,1)),x]   # for creating a matrix where we add teta_not as single column of ones

        teta=np.dot(np.linalg.inv(np.dot(x_b.T,x_b)), np.dot(x_b.T,y))   # teta=np.linalg.inv(x_b.T @ x_b) @ x_b.T @ y   (both are same formula)
        

        self.bias=teta[0]
        self.weight=teta[1:]

    def predict(self, x):
        y_pred=self.bias + np.dot(x, self.weight)
        return y_pred
```

```python
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([2,4,6,8,10])

model = LinearRegOLS()
model.fit(x,y)
y_pred=model.predict(x)
print(y_pred)
```

```python
print(model.bias)
print(model.weight)
# mean our actual equation came y_pred=2x that is why bias came 0 and teta1 mean weight here came near 2 
```

```python
# so according to results for Small datasets the OLS work far better than Gradient Decent but GD is better for huge datasets than OLS
```
