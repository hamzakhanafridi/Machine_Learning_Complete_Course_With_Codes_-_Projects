```python
import numpy as np
```

```python
class LogisticRegression:
    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.bias=None
        self.weight=None
        self.lr=learning_rate
        self.n_iter=n_iter


    def _sigmoid(self, z):
        return(1/(1+np.exp(-z)))
        
    def fit(self, x,y):

        #step 1
        m,n=x.shape
        self.bias=0
        self.weight=np.zeros(n)

        for i in range(self.n_iter):
            # Step 2
            z=self.bias + np.dot(x, self.weight)   #self.bias is teta(0) and dot is dot product and self.weight is all tetas here which multiply with x
            y_pred=self._sigmoid(z)   

            
            db=(1/m) * np.sum(y_pred - y)    # db mean derivative w.r.t teta not
            dw=(1/m) * np.dot(x.T, (y_pred - y))
    
            # Step 4 applying convergence theorem 
            self.bias -= self.lr * db
            self.weight -= self.lr * dw

    def probability(self,x):
        z=self.bias + np.dot(x, self.weight) 
        return self._sigmoid(z)   

    def predict(self, x, threshold=0.5):
        probabilites=self.probability(x)
        y_pred_bool=probabilites >= threshold
        return y_pred_bool.astype(int)
```

```python
x=np.array([[1,2],[2,3],[3,4],[4,5]])
y=np.array([0,0,1,1])

model=LogisticRegression()

model.fit(x,y)

y_pred=model.predict(x)
print(y_pred)

print(model.probability(x))
```
