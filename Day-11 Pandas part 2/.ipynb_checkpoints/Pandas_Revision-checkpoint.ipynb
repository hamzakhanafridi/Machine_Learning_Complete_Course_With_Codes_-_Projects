```python
matr=pd.DataFrame({
 'country':['pak','china','usa','dinmark','UAE'] ,  
 'year':[2020,2002,2004,1999,2925],   
 'sales':[20,2,234,99,25],   
 'Profit':['20%','02%','24%','9%','5%']   
})
```

```python
matr
```

```python
melt=matr.melt(
    ['country','year'],
    ['sales','Profit'],
    'metrix',
    'values'
)
melt
```

```python
original_pivot=melt.pivot(
    index=['country','year'],
    columns='metrix',
    values='values'
)
```

```python
original_pivot
```

```python
matr
```

```python
matr['sales'].plot()
```

```python
matr.plot(kind='scatter',x='sales',y='Profit')
```

```python
pd.concat([melt,original_pivot],ignore_index=True)  # column wise concatination
```

```python
pd.concat([melt,original_pivot])  # row wise concatination
```
