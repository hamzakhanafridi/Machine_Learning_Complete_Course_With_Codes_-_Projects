#  Lecture 4
## Now cleaning and structuring the above data


```python
def clean(data):
    clean_user = {'one':1 , 'two':2, 'three':3, 'four':4, 'five':5}

    clean_data=[]
    unique_user=set()    # this is used in removing duplicate data
    for user in data:
        raw_rating = str(user['rating']).strip().lower()

        if raw_rating in clean_user:
            raw_rating = clean_user[raw_rating]
        user['rating'] = raw_rating


#   handling missing values 
        raw_age=user.get('age')
        if raw_age == None:
            user['age']=None

        # Deduplication  (mean removing duplicate data)
        if user['name'] in unique_user:
            continue
        unique_user.add(user['name'])
        clean_data.append(user)

    return clean_data
```

```python
data=clean(data)
```

```python
print(data)
```

# Lecture no 5
## finding insights (total number of rating and poor rating percentage)

```python
# Finding the total number of average rating
def get_per(data):
    total_rat=0
    for user in data:
        total_rat += float(user['rating'])

    print(f'the total number of rating is {total_rat/len(data)}')
```

```python
print(get_per(data))
```

```python
# finding the percentage of poor rating
def get_per(data):
    poor_rat=0
    for user in data:
        if float(user['rating']) < 3:
            poor_rat +=1

    print(f'The percentage of poor rating is {poor_rat/len(data) *100}')
            
```

```python
get_per(data)
```

# Lecture 6 
## create a recommendation system on above data
- for user rating above 4 recommend them apple mobile 
- for user rating below 4 recommend them samsung mobile 

```python
def get_recommendation(data):
    recommendation=[]
    for user in data:
        current_user={}
        current_user['name']=user['name']
        if float(user['rating']) > 4:
            current_user['brand'] = 'Apple'
        else:
            current_user['brand'] = 'Samsung'

        recommendation.append(current_user)

    return recommendation
        
```

```python
print(get_recommendation(data))
```
